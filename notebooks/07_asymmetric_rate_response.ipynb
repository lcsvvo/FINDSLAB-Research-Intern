## 3단계. 상승·하락 변수 생성과 검증

### 왜 이렇게 나누나?

`d_y3`(3년물 금리변화)와 `d_ns_curvature`(곡률변화)를 각각 상승분·하락분 두 변수로 나눈다.
`clip`으로 만들면 두 변수는 겹치지 않고, 더하면 항상 원변수로 돌아온다.

| d_y3 | d_y3_up | d_y3_down |
|---:|---:|---:|
| +0.026 | +0.026 | 0.000 |
| −0.030 | 0.000 | −0.030 |
| 0.000 | 0.000 | 0.000 |

### Decision Box 1 — 금리 변화를 상승분·하락분으로 나눈다

- **문제:** 금리 변화를 하나의 변수로 볼 것인가, 상승과 하락으로 나눌 것인가.
- **선택:** Δy3를 상승분(≥0)·하락분(≤0) 두 변수로 나눠 계수를 각각 추정한다. `clip`으로 만들므로 두 변수는 겹치지 않고 더하면 원변수로 정확히 돌아온다.
- **근거:** "금리가 오를 때와 내릴 때 산업이 다르게 반응하는가"가 H2의 질문이다. 변수를 하나로 두면 두 방향의 평균만 보이므로 이 질문을 검정할 수 없다. 사전에 계획된 확증 분석이다.
- **결론에 미친 영향:** 산업별 Wald 검정(H0: 상승계수 = 하락계수) 7개 중 BH 보정 후 유의한 산업은 <b>건설업 하나뿐</b>이다(F=10.42, q=0.0089). 건설업은 금리 상승 시 −0.022, 하락 시 +0.031로 <b>부호가 반대</b>다. 7개 산업 전체에 대한 공동 대칭성 검정도 클러스터 SE 기준 <b>p=0.0401</b>로 5%에서 기각된다. 반면 곡률에서는 어느 산업도 비대칭이 아니다(공동검정 p=0.617). 즉 비대칭은 <b>금리 수준 변화에서, 그것도 건설업에 한정해</b> 나타난다.

```python
panel["d_y3_up"] = panel["d_y3"].clip(lower=0)
panel["d_y3_down"] = panel["d_y3"].clip(upper=0)
panel["d_curv_up"] = panel["d_ns_curvature"].clip(lower=0)
panel["d_curv_down"] = panel["d_ns_curvature"].clip(upper=0)

valid_y3 = panel["d_y3"].notna()
valid_curv = panel["d_ns_curvature"].notna()

# 분해 항등식: x == x_up + x_down (결측이 아닌 행에서), 2건
assert np.allclose(
    panel.loc[valid_y3, "d_y3"],
    panel.loc[valid_y3, "d_y3_up"] + panel.loc[valid_y3, "d_y3_down"],
)
assert np.allclose(
    panel.loc[valid_curv, "d_ns_curvature"],
    panel.loc[valid_curv, "d_curv_up"] + panel.loc[valid_curv, "d_curv_down"],
)

# 부호 검증, 2건
assert (panel["d_y3_up"].dropna() >= 0).all() and (panel["d_curv_up"].dropna() >= 0).all()
assert (panel["d_y3_down"].dropna() <= 0).all() and (panel["d_curv_down"].dropna() <= 0).all()

print("상승·하락 변수 생성 및 검증 완료")
display(
    panel[["date", "d_y3", "d_y3_up", "d_y3_down", "d_ns_curvature", "d_curv_up", "d_curv_down"]]
    .drop_duplicates("date")
    .head(10)
)
```

## 4단계. 표본 균형과 금리변화 기술통계

## 5단계. 산업별 비대칭 회귀 추정 (3년물)

$$
R_{i,t} = \alpha_i + \gamma_i R_{KOSPI,t} + \beta_{up,i}\Delta y_{3,t}^{up} + \beta_{down,i}\Delta y_{3,t}^{down} + \epsilon_{i,t}
$$

### 왜 KOSPI를 통제하나

- 산업수익률에는 시장 전체의 움직임이 섞여 있다. 
- 이를 통제하지 않으면 금리효과로 보이는 계수가
실제로는 시장베타에 의한 것일 수 있다. `kospi_return`을 공통 통제변수로 넣어 시장효과를 걷어낸 뒤
남는 부분만 금리 상승·하락 계수로 본다.

### HAC 표준오차

- 일별 수익률·금리변화는 시계열 자기상관을 가질 수 있어 통상 OLS 표준오차는 과소추정될 위험이 있다.
- Newey-West HAC(`maxlags=config.HAC_MAXLAGS`)를 써서 Baseline·모형 B와 표준오차 산출 방식을 통일한다.

### 하락계수 부호 읽는 법

- `d_y3_down`은 0 이하 값만 가진다. 따라서 β_down이 <b>양수</b>이면, 금리가 하락해 `d_y3_down`이 더
음수가 될수록 `β_down × d_y3_down` 항이 더 음수가 되어 산업수익률이 하락한다는 뜻이다.
- <b>β_down 양수 = 금리 하락 시 수익률 하락.</b>

```python
# add_bh_correction -- nb06(06_yield_curve_exposure.ipynb) cell 42 원문 그대로.
# PART A(5단계)에서 곧바로 BH 보정이 필요하므로 여기서 먼저 정의한다.
# (검정 함수 자체에 대한 설명은 7단계에서 run_industry_difference_test와 함께 다룬다.)
def add_bh_correction(result, group_cols=("model", "factor"),
                      p_col="p_value", alpha=0.05):
    out = result.copy()
    group_cols = list(group_cols)

    out["p_value_bh"] = np.nan
    out["bh_n_tests"] = 0

    # 길이 1 리스트로 그룹화하면 pandas 4에서 키가 스칼라 대신 튜플이 된다. 단일 열이면 스칼라로 넘긴다.
    _by = group_cols[0] if len(group_cols) == 1 else group_cols
    for _, _sub in out.groupby(_by):
        idx = _sub.index
        pvals = out.loc[idx, p_col].to_numpy(dtype=float)
        out.loc[idx, "p_value_bh"] = multipletests(pvals, method="fdr_bh")[1]
        out.loc[idx, "bh_n_tests"] = len(pvals)

    out["bh_family"] = out[group_cols].astype(str).agg(" | ".join, axis=1)
    out["significant_5pct_bh"] = out["p_value_bh"] < alpha

    return out


def fit_asymmetric(data, up, down, hac_lags=HAC_LAGS):
    # R = a + g*kospi_return + b_up*X_up + b_down*X_down (HAC)
    use_cols = ["industry_return", "kospi_return", up, down]
    sample = data[use_cols].replace([np.inf, -np.inf], np.nan).dropna()
    X = sm.add_constant(sample[["kospi_return", up, down]], has_constant="add")
    return sm.OLS(sample["industry_return"], X).fit(
        cov_type="HAC", cov_kwds={"maxlags": hac_lags}
    )


industries = sorted(panel["industry"].unique())

rows_y3 = []
for ind in industries:
    g = panel[panel["industry"] == ind]
    m = fit_asymmetric(g, "d_y3_up", "d_y3_down")
    ci = m.conf_int(alpha=0.05)
    for direction, var in [("up", "d_y3_up"), ("down", "d_y3_down")]:
        rows_y3.append({
            "model": "Y3_Asymmetry",
            "industry": ind,
            "direction": direction,
            "beta": m.params[var],
            "se_hac": m.bse[var],
            "t_value": m.tvalues[var],
            "p_value": m.pvalues[var],
            "ci_lower_95": ci.loc[var, 0],
            "ci_upper_95": ci.loc[var, 1],
            "nobs": int(m.nobs),
            "r_squared": m.rsquared,
            "hac_lags": HAC_LAGS,
        })

up_down_beta_long = add_bh_correction(
    pd.DataFrame(rows_y3), group_cols=("model", "direction")
)
display(up_down_beta_long.round(6))
```

## 7단계. 공통 검정 함수

- `add_bh_correction`은 5단계에서 정의한 것을 재사용한다(06 cell 42 원문).
- 산업 간 계수 동일성 검정 `run_industry_difference_test`는 06 cell 68 원문을 가져와, `src/models.py`의 `cross_industry_Ftest`와 결과가 일치하는지 검증한다.

### 검정 설정은 04·05·06과 동일하게 맞춘다

- 산업더미로 <b>산업별 절편</b>을 둔다.
- 산업별 시장 교차항으로 <b>시장베타도 산업마다 다르게</b> 추정한다(빠뜨리면 시장효과가 금리 교차항으로 샌다).
- <b>날짜 클러스터 표준오차</b>를 쓴다. 시계열 HAC는 같은 날 산업 간 잔차의 횡단면 상관을 놓쳐
  검정이 지나치게 보수적이 되므로, 산업 간 차이·전 산업 동시검정에는 HAC 대신 날짜 클러스터 SE를 쓴다.

```python
# run_industry_difference_test -- nb06 cell 68 원문 그대로
def run_industry_difference_test(data, test_vars, model_name,
                                 ret_col="industry_return",
                                 mkt_col="kospi_return",
                                 industry_col="industry",
                                 date_col="date"):
    """H0: 모든 산업의 계수가 같다. 요인별로 F-검정한다."""
    test_vars = list(test_vars)
    d = data.dropna(subset=[ret_col, mkt_col, industry_col, date_col] + test_vars).copy()

    date_codes = pd.factorize(d[date_col])[0]
    if len(np.unique(date_codes)) < 2:
        raise ValueError("날짜 그룹이 2개 이상이어야 클러스터 SE를 쓸 수 있습니다.")

    dummies = pd.get_dummies(d[industry_col], prefix="ind", dtype=float)
    blocks = [dummies, dummies.mul(d[mkt_col].values, axis=0).add_prefix("mkt_")]

    for v in test_vars:
        blocks.append(dummies.mul(d[v].values, axis=0).add_prefix(f"{v}__"))

    X = pd.concat(blocks, axis=1)
    names = list(X.columns)

    model = sm.OLS(d[ret_col].values, X.values).fit(
        cov_type="cluster", cov_kwds={"groups": date_codes})

    rows = []
    for v in test_vars:
        idx = [i for i, c in enumerate(names) if c.startswith(f"{v}__")]

        # "첫 산업 계수 = 나머지 산업 계수" 제약행렬
        R = np.zeros((len(idx) - 1, len(names)))
        for r, j in enumerate(idx[1:]):
            R[r, idx[0]] = 1.0
            R[r, j] = -1.0

        t = model.f_test(R)

        rows.append({
            "model": model_name,
            "variable": v,
            "f_stat": float(t.fvalue),
            "p_value": float(t.pvalue),
            "df_num": int(len(idx) - 1),
            "n_industries": int(dummies.shape[1]),
            "n_obs": int(model.nobs),
            "n_dates": int(len(np.unique(date_codes))),
            "significant_5pct": bool(t.pvalue < 0.05),
        })

    return pd.DataFrame(rows)


# 검증: 3년물 하나만 넣어 src/models.py 의 기존 함수와 결과가 같은지 확인한다.
from src.models import cross_industry_Ftest

_check_new = run_industry_difference_test(panel, ["d_y3"], "check")
f_new = float(_check_new.loc[0, "f_stat"])

_long_df = panel.rename(columns={
    "industry_return": "ret", "kospi_return": "mkt", "d_y3": "dY",
})
_ft_old = cross_industry_Ftest(_long_df, date_col="date")
f_old = float(_ft_old.fvalue)

f_diff = abs(f_new - f_old)
print(f"run_industry_difference_test F = {f_new:.10f}")
print(f"cross_industry_Ftest F         = {f_old:.10f}")
print(f"차이 = {f_diff:.3e}")

assert f_diff < 1e-8, f"F값 불일치: {f_diff}"
print("-> 일치. run_industry_difference_test는 기존 사양을 그대로 따른다.")
```

## 8단계. 3년물 비대칭 직접검정 (H0: beta_up = beta_down)

### 계수 유의성과 비대칭은 다르다

- 하락계수(또는 상승계수) 하나가 유의한 것과, beta_up과 beta_down이 서로 다르다는 것은 다른 질문이다.
- 전자는 개별 계수에 대한 t검정이고, 후자는 두 계수의 차이에 대한 Wald(F) 검정이다.
- 두 계수가 각각 유의해도 크기가 비슷하면 비대칭이 아니고, 반대로 각각은 유의하지 않아도
방향에 따라 크기가 크게 다르면 비대칭일 수 있다.

```python
rows_y3_wald = []
for ind in industries:
    g = panel[panel["industry"] == ind]
    m = fit_asymmetric(g, "d_y3_up", "d_y3_down")
    ci = m.conf_int(alpha=0.05)
    w = m.f_test("d_y3_up = d_y3_down")

    rows_y3_wald.append({
        "model": "Y3_Wald",
        "industry": ind,
        "f_stat": float(w.fvalue),
        "p_value": float(w.pvalue),
        "beta_up": m.params["d_y3_up"],
        "se_up": m.bse["d_y3_up"],
        "ci_lower_up": ci.loc["d_y3_up", 0],
        "ci_upper_up": ci.loc["d_y3_up", 1],
        "beta_down": m.params["d_y3_down"],
        "se_down": m.bse["d_y3_down"],
        "ci_lower_down": ci.loc["d_y3_down", 0],
        "ci_upper_down": ci.loc["d_y3_down", 1],
        "nobs": int(m.nobs),
        "r_squared": m.rsquared,
    })

asymmetry_wald_tests = add_bh_correction(pd.DataFrame(rows_y3_wald), group_cols=("model",))

wald_path = TABDIR / "asymmetry_wald_tests.csv"
asymmetry_wald_tests.to_csv(wald_path, index=False, encoding="utf-8-sig")
display(asymmetry_wald_tests.round(6))
```

> 이 표는 산업별 3년물 상승·하락 계수와 H0: beta_up = beta_down Wald 검정 결과를 보여준다.
> 
> - <b>SE</b>: 산업별 개별 시계열 회귀이므로 HAC(Newey–West, lag 5). 부록 A 참고.
> - <b>BH family</b>: 「Y3_Wald」 7개 산업이 한 족이다. 곡률 Wald(`Curvature_Wald`)와도 묶지 않는다 — 서로 다른 요인에 대한 질문이기 때문이다. 보정 전후 p값은 부록 B 참고.

## 8-1단계. 건설업 비대칭의 성격 — 방향인가 크기인가 (§7)

- 건설업만 비대칭이 유의했다.
- 그런데 βup = −0.0216, βdown = +0.0310은 Δydown ≤ 0을 넣으면 <b>두 항 모두 음(−)의 기여</b>를 낸다
    — 금리가 오르든 내리든 건설업 수익률은 떨어진다. 이는 "상승 충격이 하락보다 크다"는 통상적 방향 비대칭과 다르다.

- 두 항을 절댓값 성분과 방향 성분으로 다시 묶으면
    - β_up·Δy_up + β_down·Δy_down = β_abs·|Δy3| + β_dir·Δy3.
    - Δy_up − Δy_down = |Δy| 이고 Δy_up + Δy_down = Δy 이므로, β_abs = (β_up − β_down)/2, β_dir = (β_up + β_down)/2 이다.

- 건설업 추정치를 넣으면 다음과 같다(§7 피드백 분해식).
    - −0.0216·Δy_up + 0.0310·Δy_down = <b>−0.0263</b>·|Δy3| + <b>0.0047</b>·Δy3
    - 절댓값 성분이 방향 성분의 <b>약 5.6배</b>다.

- 절댓값 성분이 방향 성분보다 크면, 이 산업은 금리 방향이 아니라 <b>움직임의 크기(금리 불확실성)</b>에 반응하는 것이다. 절댓값 모형으로 직접 확인한다.

> 분해 항등식의 수치 검증(7개 산업 전부)은 `07b_rate_uncertainty_exposure.ipynb` 3단계와 `uncert_reparameterization.csv`에 있다.

```python
# 절댓값 모형: 방향(Δy3) 대신 크기(|Δy3|)에 반응하는지 검정 (§7)
import statsmodels.api as sm
panel["abs_d_y3"] = panel["d_y3_up"] - panel["d_y3_down"]      # |Δy3| = 상승분 − 하락분

def _fit_abs(g, xcols):
    r = g[["industry_return"] + xcols].dropna()
    return sm.OLS(r["industry_return"], sm.add_constant(r[xcols])).fit(
        cov_type="HAC", cov_kwds={"maxlags": config.HAC_MAXLAGS})

abs_rows = []
for _ind, _g in panel.groupby("industry"):
    _m_abs = _fit_abs(_g, ["kospi_return", "abs_d_y3"])            # 크기만
    _m_both = _fit_abs(_g, ["kospi_return", "abs_d_y3", "d_y3"])   # 크기 + 방향
    abs_rows.append({"industry": _ind,
                     "beta_abs": _m_abs.params["abs_d_y3"], "p_abs": _m_abs.pvalues["abs_d_y3"],
                     "beta_dir_both": _m_both.params["d_y3"], "p_dir_both": _m_both.pvalues["d_y3"]})
absolute_value_model = pd.DataFrame(abs_rows).sort_values("industry").reset_index(drop=True)
display(absolute_value_model.round(4))
absolute_value_model.to_csv(config.H2_TABDIR / "absolute_value_model.csv",
                            index=False, encoding="utf-8-sig")

_t = absolute_value_model.sort_values("beta_abs")
_colors = [config.sign_color(b, pv < 0.05) for b, pv in zip(_t["beta_abs"], _t["p_abs"])]
fig, ax = plt.subplots(figsize=(7, 4))
ax.barh(_t["industry"], _t["beta_abs"], color=_colors)
ax.axvline(0, color="gray", lw=1)
ax.set_xlabel("크기 계수 β_abs (|Δy3| 반응)")
ax.set_title("산업별 금리 변동크기 반응 (건설업만 유의 음)")
fig.tight_layout()
fig.savefig(config.H2_FIGDIR / "05b_absolute_value_beta.png", dpi=120, bbox_inches="tight")
plt.show()
```

### 관찰 결과

- 건설업 크기 계수는 <b>사양에 따라 두 값</b>이 나온다.
    - |Δy3| <b>단독</b> 모형: β_abs = −0.0258 (p≈0.002, 유의) — 아래 표의 `beta_abs` 열.
    - |Δy3|와 Δy3를 <b>함께</b> 넣은 모형: β_abs = −0.0263 — §7 피드백 분해식의 값.
    - 두 값의 차이(0.0005)는 방향 계수가 작아서 생기는 사양 차이이며, 결론을 바꾸지 않는다. 다만 <b>분해식을 인용할 때는 −0.0263을 쓴다</b>.
- 함께 넣은 모형에서 방향 계수는 +0.0047로 <b>비유의</b>(p≈0.56)다. 절댓값 성분이 방향 성분의 약 5.6배다.
- 즉 건설업은 금리 방향이 아니라 <b>변동 크기</b>에 반응한다 — 금리 불확실성 반응.
- 나머지 산업은 반대로 방향 계수가 유의하다. 크기 반응이 건설업에 한정되는지에 대한 산업 간 검정은 이 STEP에서 수행하지 않았다(`07b` 5단계 참고).

### 해석 (§7)

- 건설업의 "비대칭"은 상승 충격이 큰 게 아니라 금리가 어느 쪽으로 움직이든 하락하는 <b>변동성 반응</b>이다. 부채구조·분양 일정이 금리 경로 가정에 묶여 있어(실물옵션) 불확실성에 민감하다는 해석과 부합한다.
- 설명변수를 둘에서 하나(|Δy3|)로 줄이면서 같은 사실을 말할 수 있어 보고가 간결해진다.

## 9단계. 후속 점검 요인(곡률) 계수 추정

### 왜 곡률인가

- 모형 B(`06_yield_curve_exposure.ipynb`)에서는 Level·Slope·Curvature 세 요인 중
<b>곡률만</b> 산업 간 차이 검정을 통과했다(BH p=0.0047). 
- 그래서 이 STEP의
후속 점검 요인은 λ=0.5 사양의 곡률로 한정한다. 대체 λ·대체 곡률에서 H1의 산업 간 차이가 유지되지 않으므로, 이 선택은 일반적 강건성의 주장이 아니라 H1 사양을 H2에서 연결해 확인하기 위한 것이다. 
- λ는 모형 B와 동일하게 `config.NS_LAMBDA`(0.5)를 쓴다.

```python
rows_curv_beta = []
for ind in industries:
    g = panel[panel["industry"] == ind]
    m = fit_asymmetric(g, "d_curv_up", "d_curv_down")
    ci = m.conf_int(alpha=0.05)
    for direction, var in [("up", "d_curv_up"), ("down", "d_curv_down")]:
        rows_curv_beta.append({
            "model": "Curvature_Coef",
            "direction": direction,
            "industry": ind,
            "beta": m.params[var],
            "se_hac": m.bse[var],
            "p_value": m.pvalues[var],
            "ci_lower_95": ci.loc[var, 0],
            "ci_upper_95": ci.loc[var, 1],
            "nobs": int(m.nobs),
            "r_squared": m.rsquared,
        })

key_factor_asymmetry_tests = add_bh_correction(
    pd.DataFrame(rows_curv_beta), group_cols=("model", "direction")
)

curv_path = TABDIR / "key_factor_asymmetry_tests.csv"
key_factor_asymmetry_tests.to_csv(curv_path, index=False, encoding="utf-8-sig")
display(key_factor_asymmetry_tests.round(6))
```

> - <b>BH family</b>: 「Curvature_Coef | up」 7개와 「Curvature_Coef | down」 7개를 각각 별도 족으로 보정했다. 보정 전후 p값은 부록 B 참고.

## 10단계. 곡률 비대칭 직접검정 (H0: beta_curv_up = beta_curv_down)

곡률 비대칭은 <b>사전 계획된 검정</b>이며, 산업별·전 산업 동시검정 모두 기각에 실패한다(유의성 없음). 유의하지 않아도 계획된 검정이므로 본문에 보고한다(§4.1).

```python
rows_curv_wald = []
for ind in industries:
    g = panel[panel["industry"] == ind]
    m = fit_asymmetric(g, "d_curv_up", "d_curv_down")
    ci = m.conf_int(alpha=0.05)
    w = m.f_test("d_curv_up = d_curv_down")

    rows_curv_wald.append({
        "model": "Curvature_Wald",
        "industry": ind,
        "f_stat": float(w.fvalue),
        "p_value": float(w.pvalue),
        "beta_up": m.params["d_curv_up"],
        "se_up": m.bse["d_curv_up"],
        "ci_lower_up": ci.loc["d_curv_up", 0],
        "ci_upper_up": ci.loc["d_curv_up", 1],
        "beta_down": m.params["d_curv_down"],
        "se_down": m.bse["d_curv_down"],
        "ci_lower_down": ci.loc["d_curv_down", 0],
        "ci_upper_down": ci.loc["d_curv_down", 1],
        "nobs": int(m.nobs),
        "r_squared": m.rsquared,
    })

key_factor_asymmetry_wald_tests = add_bh_correction(pd.DataFrame(rows_curv_wald), group_cols=("model",))

curv_wald_path = TABDIR / "key_factor_asymmetry_wald_tests.csv"
key_factor_asymmetry_wald_tests.to_csv(curv_wald_path, index=False, encoding="utf-8-sig")
display(key_factor_asymmetry_wald_tests.round(6))
```

> 이 표는 산업별 곡률 상승·하락 계수와 H0: beta_curv_up = beta_curv_down Wald 검정 결과를 보여준다.
> 
> - <b>SE</b>: 산업별 개별 시계열 회귀이므로 HAC(Newey–West, lag 5). 부록 A 참고.
> - <b>BH family</b>: 「Curvature_Wald」 7개 산업이 한 족이다. 보정 전후 p값은 부록 B 참고.

## 12단계. 방향 분리의 정당성 — 전 산업 동시검정

### 산업 간 차이 검정으로는 답할 수 없다

(a) 11단계의 산업 간 차이 검정은 "산업마다 다른가"를 묻는 별개의 질문이며,
"상승과 하락이 다른가"라는 방향 분리 정당성에 대한 답이 아니다.

(b) 대칭 모형과 방향 분리 모형의 F값을 단순 비교하는 것도 근거가 되지 못한다.
- 방향을 나누면 각 계수가 전체 관측치의 절반(상승일 또는 하락일)만으로 추정되어
표준오차가 커진다(아래에서 실제로 약 몇 배인지 계산한다). 
- 방향 분리 모형의 F가
낮아지는 것은 이 검정력 감소만으로도 자동으로 나타날 수 있는 현상이므로,
F값 비교 자체가 방향 분리를 지지하거나 반박하는 근거가 될 수 없다.

그래서 방향 분리가 정당한지는 <b>전 산업을 대상으로 한 번에</b> H0: 모든 산업 i 에서
beta_up_i = beta_down_i 를 검정해서 직접 답한다(BH 보정 없음 — 모형당 단일 검정).

```python
def joint_symmetry_test(data, up, down, date_col="date"):
    """H0: 모든 산업 i 에 대해 b_up_i = b_down_i (7개 제약). 날짜 클러스터 SE."""
    d = data.dropna(subset=["industry", "industry_return", "kospi_return",
                            date_col, up, down]).copy()
    date_codes = pd.factorize(d[date_col])[0]

    dummies = pd.get_dummies(d["industry"], prefix="ind", dtype=float)
    X = pd.concat([
        dummies,
        dummies.mul(d["kospi_return"].values, axis=0).add_prefix("mkt_"),
        dummies.mul(d[up].values, axis=0).add_prefix(f"{up}__"),
        dummies.mul(d[down].values, axis=0).add_prefix(f"{down}__"),
    ], axis=1)

    res = sm.OLS(d["industry_return"].values, X.values).fit(
        cov_type="cluster", cov_kwds={"groups": date_codes})
    names = list(X.columns)
    iu = [i for i, c in enumerate(names) if c.startswith(f"{up}__")]
    idn = [i for i, c in enumerate(names) if c.startswith(f"{down}__")]

    R = np.zeros((len(iu), len(names)))
    for k, (a, b) in enumerate(zip(iu, idn)):
        R[k, a], R[k, b] = 1.0, -1.0

    return res.f_test(R), res, iu, idn


def sym_industry_model(data, var, date_col="date"):
    """대칭 모형(변수 1개, 산업 교차항)의 교차항 평균 클러스터 SE."""
    d = data.dropna(subset=["industry", "industry_return", "kospi_return", date_col, var]).copy()
    date_codes = pd.factorize(d[date_col])[0]
    dummies = pd.get_dummies(d["industry"], prefix="ind", dtype=float)
    X = pd.concat([
        dummies,
        dummies.mul(d["kospi_return"].values, axis=0).add_prefix("mkt_"),
        dummies.mul(d[var].values, axis=0).add_prefix(f"{var}__"),
    ], axis=1)
    res = sm.OLS(d["industry_return"].values, X.values).fit(
        cov_type="cluster", cov_kwds={"groups": date_codes})
    names = list(X.columns)
    idx = [i for i, c in enumerate(names) if c.startswith(f"{var}__")]
    return float(res.bse[idx].mean())


joint_y3, res_y3, iu_y3, idn_y3 = joint_symmetry_test(panel, "d_y3_up", "d_y3_down")
joint_curv, res_curv, iu_curv, idn_curv = joint_symmetry_test(panel, "d_curv_up", "d_curv_down")

joint_symmetry_tests = pd.DataFrame([
    {
        "model": "Y3_Asymmetry",
        "hypothesis": "모든 산업에서 d_y3_up = d_y3_down",
        "f_stat": float(joint_y3.fvalue),
        "df_num": int(joint_y3.df_num),
        "p_value": float(joint_y3.pvalue),
        "se_type": "cluster(date)",
        "reject_5pct": bool(float(joint_y3.pvalue) < 0.05),
    },
    {
        "model": "Curvature_Asymmetry",
        "hypothesis": "모든 산업에서 d_curv_up = d_curv_down",
        "f_stat": float(joint_curv.fvalue),
        "df_num": int(joint_curv.df_num),
        "p_value": float(joint_curv.pvalue),
        "se_type": "cluster(date)",
        "reject_5pct": bool(float(joint_curv.pvalue) < 0.05),
    },
])

joint_path = TABDIR / "joint_symmetry_tests.csv"
joint_symmetry_tests.to_csv(joint_path, index=False, encoding="utf-8-sig")

se_sym_y3 = sym_industry_model(panel, "d_y3")
se_dir_y3 = float(np.mean(np.concatenate([res_y3.bse[iu_y3], res_y3.bse[idn_y3]])))
se_sym_curv = sym_industry_model(panel, "d_ns_curvature")
se_dir_curv = float(np.mean(np.concatenate([res_curv.bse[iu_curv], res_curv.bse[idn_curv]])))

print(f"3년물 교차항 평균 SE: 대칭 {se_sym_y3:.5f} -> 방향분리 {se_dir_y3:.5f} "
      f"(x{se_dir_y3/se_sym_y3:.2f})")
print(f"곡률 교차항 평균 SE: 대칭 {se_sym_curv:.5f} -> 방향분리 {se_dir_curv:.5f} "
      f"(x{se_dir_curv/se_sym_curv:.2f})")
display(joint_symmetry_tests.round(6))
```

> 이 표는 전 산업을 대상으로 한 번에 검정하는 방향 분리 정당성 Wald 검정(H0: 모든 산업에서 beta_up=beta_down, 제약 7개)을 보여준다.
> 
> - <b>SE</b>: 7개 산업이 한 식에 들어가는 통합회귀이므로 날짜 클러스터. 부록 A 참고.
> - <b>BH family</b>: 모형당 검정이 1개뿐이라 보정할 대상이 없다. BH를 적용하지 않고 원 p값을 그대로 보고한다.

```python
wald_bh_map = {
    "y3": asymmetry_wald_tests.set_index("industry")["p_value_bh"].to_dict(),
    "curv": key_factor_asymmetry_wald_tests.set_index("industry")["p_value_bh"].to_dict(),
}


def panel_ci(ax, beta_df, wald_map, title, xlabel):
    """산업별 상승·하락 계수와 95% CI를 그린다.

    색상은 계수 부호를 나타낸다. 점·에러바 색은 config.sign_color(beta, significant)
    로 정해 '부호'를 인코딩한다(유의 기준: BH 보정 p<0.05). 상승/하락 구분은 색이 아니라
    마커 모양(o=상승, s=하락)과 세로 오프셋으로만 표현한다.
    """
    piv_up = beta_df[beta_df["direction"] == "up"].set_index("industry")
    piv_down = beta_df[beta_df["direction"] == "down"].set_index("industry")
    order = sorted(beta_df["industry"].unique(), reverse=True)
    y = np.arange(len(order))

    for i, ind in enumerate(order):
        bu, lu, hu, pu = (piv_up.loc[ind, c] for c in ("beta", "ci_lower_95", "ci_upper_95", "significant_5pct_bh"))
        bd, ld, hd, pd_ = (piv_down.loc[ind, c] for c in ("beta", "ci_lower_95", "ci_upper_95", "significant_5pct_bh"))

        ax.plot([bd, bu], [y[i], y[i]], color="#cbd5e0", lw=3, alpha=0.7, zorder=1)

        for beta, lo, hi, sig, off, mk in [
            (bu, lu, hu, pu, +0.16, "o"),
            (bd, ld, hd, pd_, -0.16, "s"),
        ]:
            col = config.sign_color(beta, sig)
            ax.errorbar(beta, y[i] + off, xerr=[[beta - lo], [hi - beta]],
                        fmt=mk, ms=7, capsize=3, lw=1.4,
                        color=col, ecolor=col,
                        mfc=col if sig else "white",
                        mec=col, zorder=3)

    ax.axvline(0, color="black", lw=0.9, ls="--", alpha=0.7)
    ax.set_yticks(y)
    ax.set_yticklabels([
        f"{name} \u2605" if wald_map.get(name, 1.0) < 0.05 else name
        for name in order
    ])
    ax.set_ylim(-0.5, len(order) - 0.5)
    ax.set_title(title, fontsize=12.5, fontweight="bold", pad=10)
    ax.set_xlabel(xlabel, fontsize=10)
    ax.grid(axis="x", ls=":", color="gray", alpha=0.45)
    ax.set_axisbelow(True)
    for s in ("top", "right"):
        ax.spines[s].set_visible(False)


fig, axes = plt.subplots(1, 2, figsize=(14.5, 6.4))
panel_ci(axes[0], up_down_beta_long, wald_bh_map["y3"], "국고채 3년물", "금리베타 (수익률 소수 / 금리 %p)")
panel_ci(axes[1], key_factor_asymmetry_tests, wald_bh_map["curv"], "수익률곡선 곡률 (lambda=0.5)", "곡률베타 (수익률 소수 / 곡률 %p)")

color_handles = [
    plt.Line2D([], [], color=config.SIGN_COLORS["neg"], marker="o", ls="", ms=8, label="유의한 음(\u2212)"),
    plt.Line2D([], [], color=config.SIGN_COLORS["pos"], marker="o", ls="", ms=8, label="유의한 양(+)"),
    plt.Line2D([], [], color=config.SIGN_COLORS["ns"], marker="o", ls="", ms=8, mfc="white", label="비유의"),
]
shape_handles = [
    plt.Line2D([], [], color="black", marker="o", ls="", ms=7, label="상승계수 (o)"),
    plt.Line2D([], [], color="black", marker="s", ls="", ms=7, label="하락계수 (s)"),
    plt.Line2D([], [], color="none", label="\u2605 Wald 검정 beta_up != beta_down (BH p<0.05)"),
]
color_legend = fig.legend(handles=color_handles, loc="lower center", ncol=3, frameon=False,
                           title="색 = 부호·유의성", fontsize=9.5, title_fontsize=9.5,
                           bbox_to_anchor=(0.28, -0.12))
fig.legend(handles=shape_handles, loc="lower center", ncol=1, frameon=False,
           title="모양 = 방향", fontsize=9.5, title_fontsize=9.5,
           bbox_to_anchor=(0.8, -0.14))
fig.suptitle("산업별 금리 상승·하락 계수와 HAC 95% 신뢰구간", fontsize=15, fontweight="bold", y=0.99)
fig.tight_layout(rect=(0, 0.14, 1, 0.96))

fig02_path = FIGDIR / "02_up_down_beta_ci.png"
fig.savefig(fig02_path, dpi=200, bbox_inches="tight")
plt.show()
```

## D-1. 공통 함수

### 산업별 비대칭 Wald 검정

각 산업에서 다음을 검정한다.

\[
H_0:\beta_{up,i}=\beta_{down,i}
\]

### 전 산업 공동검정

7개 산업의 방향 차이를 한 번에 검정한다.

\[
H_0:\beta_{up,i}=\beta_{down,i}\quad \text{for all }i
\]

- `date_cluster`: 같은 날짜의 산업 간 오차 상관을 고려
- `driscoll_kraay`: 같은 날짜의 횡단면 상관과 날짜 간 시계열 의존성을 함께 고려

```python
# ============================================================
# 3단계. 공통 함수
# ============================================================

def add_bh_correction(
    result,
    group_cols=("model",),
    p_col="p_value",
    alpha=BH_ALPHA,
):
    """지정한 검정족별로 BH 보정을 적용한다."""
    out = result.copy()
    group_cols = list(group_cols)

    out["p_value_bh"] = np.nan
    out["bh_n_tests"] = 0

    # 길이 1 리스트로 그룹화하면 pandas 4에서 키가 스칼라 대신 튜플이 된다. 단일 열이면 스칼라로 넘긴다.
    _by = group_cols[0] if len(group_cols) == 1 else group_cols
    for _, _sub in out.groupby(
        _by,
        dropna=False,
    ):
        idx = _sub.index

        pvals = out.loc[
            idx,
            p_col,
        ].to_numpy(dtype=float)

        out.loc[idx, "p_value_bh"] = multipletests(
            pvals,
            alpha=alpha,
            method="fdr_bh",
        )[1]

        out.loc[idx, "bh_n_tests"] = len(pvals)

    out["bh_family"] = (
        out[group_cols]
        .astype(str)
        .agg(" | ".join, axis=1)
    )

    out["significant_5pct_bh"] = (
        out["p_value_bh"] < alpha
    )

    return out


def fit_asymmetric(
    data,
    up_col,
    down_col,
    hac_lags=HAC_LAGS,
    include_direction_dummy=False,
    drop_zero_for_dummy=True,
):
    """산업 하나의 상승·하락 비대칭 회귀를 추정한다."""

    use_cols = [
        "industry_return",
        "kospi_return",
        up_col,
        down_col,
    ]

    sample = (
        data[use_cols]
        .replace([np.inf, -np.inf], np.nan)
        .dropna()
        .copy()
    )

    x_cols = [
        "kospi_return",
        up_col,
        down_col,
    ]

    if include_direction_dummy:

        if drop_zero_for_dummy:
            sample = sample[
                (sample[up_col] > 0)
                | (sample[down_col] < 0)
            ].copy()

        sample["direction_up_dummy"] = (
            sample[up_col] > 0
        ).astype(float)

        x_cols = [
            "kospi_return",
            "direction_up_dummy",
            up_col,
            down_col,
        ]

    X = sm.add_constant(
        sample[x_cols],
        has_constant="add",
    )

    return sm.OLS(
        sample["industry_return"],
        X,
    ).fit(
        cov_type="HAC",
        cov_kwds={
            "maxlags": hac_lags,
        },
    )


def industry_wald_table(
    data,
    up_col,
    down_col,
    model_name,
    include_direction_dummy=False,
    drop_zero_for_dummy=True,
):
    """
    7개 산업의 계수와
    H0: beta_up = beta_down
    Wald 검정을 계산한다.
    """

    rows = []

    industries = sorted(
        data["industry"]
        .dropna()
        .unique()
    )

    for industry in industries:

        group = data[
            data["industry"] == industry
        ]

        model = fit_asymmetric(
            group,
            up_col,
            down_col,
            include_direction_dummy=include_direction_dummy,
            drop_zero_for_dummy=drop_zero_for_dummy,
        )

        ci = model.conf_int(alpha=0.05)

        wald = model.f_test(
            f"{up_col} = {down_col}"
        )

        row = {
            "model": model_name,
            "industry": industry,

            "beta_up": model.params[up_col],
            "se_up": model.bse[up_col],
            "ci_lower_up": ci.loc[up_col, 0],
            "ci_upper_up": ci.loc[up_col, 1],
            "p_up": model.pvalues[up_col],

            "beta_down": model.params[down_col],
            "se_down": model.bse[down_col],
            "ci_lower_down": ci.loc[down_col, 0],
            "ci_upper_down": ci.loc[down_col, 1],
            "p_down": model.pvalues[down_col],

            "beta_difference": (
                model.params[up_col]
                - model.params[down_col]
            ),

            "f_stat": float(
                np.asarray(
                    wald.fvalue
                ).squeeze()
            ),

            "p_value": float(
                np.asarray(
                    wald.pvalue
                ).squeeze()
            ),

            "nobs": int(model.nobs),
            "r_squared": model.rsquared,

            "direction_dummy_included":
                include_direction_dummy,

            "direction_dummy_beta": (
                model.params["direction_up_dummy"]
                if "direction_up_dummy"
                in model.params.index
                else np.nan
            ),

            "direction_dummy_p": (
                model.pvalues["direction_up_dummy"]
                if "direction_up_dummy"
                in model.params.index
                else np.nan
            ),
        }

        rows.append(row)

    result = pd.DataFrame(rows)

    return add_bh_correction(
        result,
        group_cols=("model",),
        p_col="p_value",
    )


def coefficient_table(
    data,
    up_col,
    down_col,
    model_name,
):
    """상승·하락 개별 계수의 p값을 14행으로 반환한다."""

    rows = []

    industries = sorted(
        data["industry"]
        .dropna()
        .unique()
    )

    for industry in industries:

        group = data[
            data["industry"] == industry
        ]

        model = fit_asymmetric(
            group,
            up_col,
            down_col,
        )

        ci = model.conf_int(alpha=0.05)

        for direction, variable in [
            ("up", up_col),
            ("down", down_col),
        ]:
            rows.append(
                {
                    "model": model_name,
                    "direction": direction,
                    "industry": industry,
                    "variable": variable,
                    "beta": model.params[variable],
                    "se": model.bse[variable],
                    "p_value": model.pvalues[variable],
                    "ci_lower_95": ci.loc[variable, 0],
                    "ci_upper_95": ci.loc[variable, 1],
                }
            )

    return pd.DataFrame(rows)


def _build_pooled_design(
    data,
    up_col,
    down_col,
):
    """전 산업 공동검정 설계행렬과 제약행렬을 만든다."""

    required = [
        "date",
        "industry",
        "industry_return",
        "kospi_return",
        up_col,
        down_col,
    ]

    d = (
        data[required]
        .replace([np.inf, -np.inf], np.nan)
        .dropna()
        .sort_values(["date", "industry"])
        .copy()
    )

    dummies = pd.get_dummies(
        d["industry"],
        prefix="ind",
        dtype=float,
    )

    X = pd.concat(
        [
            dummies,

            dummies.mul(
                d["kospi_return"].to_numpy(),
                axis=0,
            ).add_prefix("mkt_"),

            dummies.mul(
                d[up_col].to_numpy(),
                axis=0,
            ).add_prefix(f"{up_col}__"),

            dummies.mul(
                d[down_col].to_numpy(),
                axis=0,
            ).add_prefix(f"{down_col}__"),
        ],
        axis=1,
    )

    names = list(X.columns)

    up_idx = [
        i
        for i, name in enumerate(names)
        if name.startswith(f"{up_col}__")
    ]

    down_idx = [
        i
        for i, name in enumerate(names)
        if name.startswith(f"{down_col}__")
    ]

    assert (
        len(up_idx)
        == len(down_idx)
        == d["industry"].nunique()
    )

    R = np.zeros(
        (
            len(up_idx),
            len(names),
        )
    )

    for row, (u_idx, d_idx) in enumerate(
        zip(up_idx, down_idx)
    ):
        R[row, u_idx] = 1.0
        R[row, d_idx] = -1.0

    date_codes = pd.factorize(
        d["date"],
        sort=True,
    )[0]

    return d, X, R, date_codes


def joint_symmetry_test(
    data,
    up_col,
    down_col,
    variable_name,
    covariance="date_cluster",
    maxlags=HAC_LAGS,
):
    """
    모든 산업에서
    H0: beta_up = beta_down
    을 공동검정한다.
    """

    d, X, R, date_codes = _build_pooled_design(
        data,
        up_col,
        down_col,
    )

    base_model = sm.OLS(
        d["industry_return"].to_numpy(),
        X.to_numpy(),
    )

    if covariance == "date_cluster":

        result = base_model.fit(
            cov_type="cluster",
            cov_kwds={
                "groups": date_codes,
                "use_correction": True,
            },
        )

        covariance_label = "Date-cluster"

    elif covariance == "driscoll_kraay":

        result = base_model.fit(
            cov_type="hac-groupsum",
            cov_kwds={
                "time": date_codes,
                "maxlags": maxlags,
                "use_correction": "cluster",

                # Driscoll-Kraay p값이 NaN이 되는
                # 자유도 보정 문제를 방지
                "df_correction": False,
            },
        )

        covariance_label = (
            f"Driscoll-Kraay lag={maxlags}"
        )

    else:
        raise ValueError(
            "지원하지 않는 covariance입니다."
        )

    wald = result.f_test(R)

    f_stat = float(
        np.asarray(
            wald.fvalue
        ).squeeze()
    )

    p_value = float(
        np.asarray(
            wald.pvalue
        ).squeeze()
    )

    if not np.isfinite(p_value):
        raise ValueError(
            f"{covariance_label}: "
            "공동검정 p-value를 계산하지 못했습니다."
        )

    return {
        "variable": variable_name,
        "covariance": covariance,
        "covariance_label": covariance_label,

        "maxlags": (
            np.nan
            if covariance == "date_cluster"
            else maxlags
        ),

        "n_restrictions": R.shape[0],
        "f_stat": f_stat,
        "p_value": p_value,
        "nobs": int(result.nobs),
        "n_dates": int(d["date"].nunique()),
    }


def extract_industry_row(
    table,
    industry="건설업",
):
    """결과표에서 특정 산업 한 행을 가져온다."""

    match = table[
        table["industry"] == industry
    ]

    if len(match) != 1:
        raise ValueError(
            f"{industry} 행이 정확히 "
            f"1개가 아닙니다: {len(match)}개"
        )

    return match.iloc[0]


print("3단계 공통 함수 정의 완료")
```

## D-3. 극단적인 금리변화일 검증

극단값은 회귀선 기울기에 큰 영향을 줄 수 있다. 다음 사양을 비교한다.

1. 원자료
2. 금리변화 1%·99% 분위수 Winsorize
3. 절대 금리변화 상위 1% 날짜 제외
4. 절대 금리변화 상위 0.5% 날짜 제외
5. 절대 금리변화가 큰 상위 10개 날짜를 한 번에 하나씩 제외

여러 사양에서 건설업 계수의 방향과 Wald 결과가 비슷하면 극단값 의존성이 낮다고 본다.

```python
def add_split_columns(data, source_col, prefix):
    out = data.copy()
    out[f"{prefix}_up"] = out[source_col].clip(lower=0)
    out[f"{prefix}_down"] = out[source_col].clip(upper=0)
    return out


def run_specification(data, spec_name, up_col, down_col):
    result = industry_wald_table(
        data,
        up_col,
        down_col,
        model_name=spec_name,
    )
    result["specification"] = spec_name
    return result


extreme_spec_results = []

extreme_spec_results.append(
    run_specification(
        panel,
        "baseline",
        "d_y3_up",
        "d_y3_down",
    )
)

q01, q99 = panel["d_y3"].dropna().quantile([0.01, 0.99])
winsor_panel = panel.copy()
winsor_panel["d_y3_winsor_1pct"] = winsor_panel["d_y3"].clip(q01, q99)
winsor_panel = add_split_columns(
    winsor_panel,
    "d_y3_winsor_1pct",
    "d_y3_winsor_1pct",
)
extreme_spec_results.append(
    run_specification(
        winsor_panel,
        "winsor_1pct",
        "d_y3_winsor_1pct_up",
        "d_y3_winsor_1pct_down",
    )
)

daily_abs = (
    panel[["date", "d_y3"]]
    .drop_duplicates("date")
    .dropna()
    .assign(abs_d_y3=lambda x: x["d_y3"].abs())
)

abs_q99 = daily_abs["abs_d_y3"].quantile(0.99)
trim_dates_1pct = daily_abs.loc[
    daily_abs["abs_d_y3"] > abs_q99,
    "date",
]
trim_panel_1pct = panel[~panel["date"].isin(trim_dates_1pct)].copy()
extreme_spec_results.append(
    run_specification(
        trim_panel_1pct,
        "trim_abs_top_1pct",
        "d_y3_up",
        "d_y3_down",
    )
)

abs_q995 = daily_abs["abs_d_y3"].quantile(0.995)
trim_dates_05pct = daily_abs.loc[
    daily_abs["abs_d_y3"] > abs_q995,
    "date",
]
trim_panel_05pct = panel[~panel["date"].isin(trim_dates_05pct)].copy()
extreme_spec_results.append(
    run_specification(
        trim_panel_05pct,
        "trim_abs_top_0_5pct",
        "d_y3_up",
        "d_y3_down",
    )
)

extreme_all = pd.concat(extreme_spec_results, ignore_index=True)
extreme_construction = (
    extreme_all[extreme_all["industry"] == "건설업"]
    .sort_values("specification")
    .reset_index(drop=True)
)

extreme_all.to_csv(
    ROBUST_TABDIR / "robustness_extreme_value_all_industries.csv",
    index=False,
    encoding="utf-8-sig",
)
extreme_construction.to_csv(
    ROBUST_TABDIR / "robustness_extreme_value_construction.csv",
    index=False,
    encoding="utf-8-sig",
)

print("극단값 사양별 건설업 결과")
display(
    extreme_construction[
        [
            "specification",
            "beta_up",
            "beta_down",
            "f_stat",
            "p_value",
            "p_value_bh",
            "nobs",
        ]
    ].round(6)
)

top_extreme_dates = (
    daily_abs.nlargest(10, "abs_d_y3")
    [["date", "d_y3", "abs_d_y3"]]
    .reset_index(drop=True)
)

leave_one_out_rows = []

for _, extreme_day in top_extreme_dates.iterrows():
    reduced = panel[panel["date"] != extreme_day["date"]].copy()
    result = industry_wald_table(
        reduced,
        "d_y3_up",
        "d_y3_down",
        model_name=f"drop_{extreme_day['date'].date()}",
    )
    construction = extract_industry_row(result)

    leave_one_out_rows.append(
        {
            "excluded_date": extreme_day["date"],
            "excluded_d_y3": extreme_day["d_y3"],
            "excluded_abs_d_y3": extreme_day["abs_d_y3"],
            "beta_up": construction["beta_up"],
            "beta_down": construction["beta_down"],
            "f_stat": construction["f_stat"],
            "p_value": construction["p_value"],
            "p_value_bh": construction["p_value_bh"],
            "nobs": construction["nobs"],
        }
    )

leave_one_out_extreme = pd.DataFrame(leave_one_out_rows)
leave_one_out_extreme.to_csv(
    ROBUST_TABDIR / "robustness_leave_one_out_extreme_dates.csv",
    index=False,
    encoding="utf-8-sig",
)

print("극단적인 날짜를 하나씩 제외한 건설업 결과")
display(round_numeric(leave_one_out_extreme))
```

## D-4. 기간 분할·기간 제외·Rolling Window 검증

고정 기간과 3년 단위 이동창을 사용해 건설업 결과가 특정 시기에만 나타나는지 확인한다.

짧은 기간에서는 검정력이 낮아지므로 p값뿐 아니라 계수의 부호·크기·신뢰구간도 함께 본다.

```python
def subset_by_dates(
    data,
    start=None,
    end=None,
    exclude_start=None,
    exclude_end=None,
):
    out = data.copy()

    if start is not None:
        out = out[out["date"] >= pd.Timestamp(start)]
    if end is not None:
        out = out[out["date"] <= pd.Timestamp(end)]
    if exclude_start is not None and exclude_end is not None:
        mask = out["date"].between(
            pd.Timestamp(exclude_start),
            pd.Timestamp(exclude_end),
            inclusive="both",
        )
        out = out[~mask]

    return out.copy()


period_specs = {
    "full_2016_2025": panel,
    "split_2016_2020": subset_by_dates(
        panel, start="2016-01-01", end="2020-12-31"
    ),
    "split_2021_2025": subset_by_dates(
        panel, start="2021-01-01", end="2025-12-31"
    ),
    "period_2016_2019": subset_by_dates(
        panel, start="2016-01-01", end="2019-12-31"
    ),
    "period_2020_2021": subset_by_dates(
        panel, start="2020-01-01", end="2021-12-31"
    ),
    "period_2022_2025": subset_by_dates(
        panel, start="2022-01-01", end="2025-12-31"
    ),
    "exclude_2020_2021": subset_by_dates(
        panel, exclude_start="2020-01-01", exclude_end="2021-12-31"
    ),
    "exclude_2022_2023": subset_by_dates(
        panel, exclude_start="2022-01-01", exclude_end="2023-12-31"
    ),
}

period_results = []

for spec_name, spec_data in period_specs.items():
    if spec_data["date"].nunique() < 100:
        warnings.warn(f"{spec_name}: 날짜 수가 100일 미만이라 건너뜁니다.")
        continue

    result = industry_wald_table(
        spec_data,
        "d_y3_up",
        "d_y3_down",
        model_name=spec_name,
    )
    result["date_min"] = spec_data["date"].min()
    result["date_max"] = spec_data["date"].max()
    result["n_dates"] = spec_data["date"].nunique()
    period_results.append(result)

period_all = pd.concat(period_results, ignore_index=True)
period_construction = (
    period_all[period_all["industry"] == "건설업"]
    .sort_values("model")
    .reset_index(drop=True)
)

period_all.to_csv(
    ROBUST_TABDIR / "robustness_period_split_all_industries.csv",
    index=False,
    encoding="utf-8-sig",
)
period_construction.to_csv(
    ROBUST_TABDIR / "robustness_period_split_construction.csv",
    index=False,
    encoding="utf-8-sig",
)

print("고정 기간 사양별 건설업 결과")
display(
    round_numeric(
        period_construction[
            [
                "model",
                "date_min",
                "date_max",
                "n_dates",
                "beta_up",
                "beta_down",
                "f_stat",
                "p_value",
                "p_value_bh",
            ]
        ]
    )
)

rolling_rows = []
min_year = int(panel["date"].dt.year.min())
max_year = int(panel["date"].dt.year.max())
window_years = 3

for start_year in range(min_year, max_year - window_years + 2):
    end_year = start_year + window_years - 1
    rolling_data = subset_by_dates(
        panel,
        start=f"{start_year}-01-01",
        end=f"{end_year}-12-31",
    )

    if rolling_data["date"].nunique() < 100:
        continue

    result = industry_wald_table(
        rolling_data,
        "d_y3_up",
        "d_y3_down",
        model_name=f"rolling_{start_year}_{end_year}",
    )
    construction = extract_industry_row(result)

    rolling_rows.append(
        {
            "window": f"{start_year}-{end_year}",
            "start_year": start_year,
            "end_year": end_year,
            "n_dates": rolling_data["date"].nunique(),
            "beta_up": construction["beta_up"],
            "se_up": construction["se_up"],
            "ci_lower_up": construction["ci_lower_up"],
            "ci_upper_up": construction["ci_upper_up"],
            "beta_down": construction["beta_down"],
            "se_down": construction["se_down"],
            "ci_lower_down": construction["ci_lower_down"],
            "ci_upper_down": construction["ci_upper_down"],
            "f_stat": construction["f_stat"],
            "p_value": construction["p_value"],
            "p_value_bh": construction["p_value_bh"],
        }
    )

rolling_construction = pd.DataFrame(rolling_rows)
rolling_construction.to_csv(
    ROBUST_TABDIR / "robustness_rolling_3y_construction.csv",
    index=False,
    encoding="utf-8-sig",
)

print("3년 Rolling Window 건설업 결과")
display(round_numeric(rolling_construction))
```

## D-5. 전 산업 공동검정의 표준오차 비교

기존 공동검정은 날짜 클러스터 표준오차를 사용했다. 이는 같은 날짜의 산업 간 상관을 허용하지만 서로 다른 날짜 사이의 자기상관을 직접 반영하지 않는다.

이번에는 다음을 비교한다.

1. 날짜 클러스터
2. Driscoll–Kraay lag 1
3. Driscoll–Kraay lag 5
4. Driscoll–Kraay lag 10

```python
joint_rows = []

for variable_name, up_col, down_col in [
    ("Y3", "d_y3_up", "d_y3_down"),
    ("Curvature", "d_curv_up", "d_curv_down"),
]:
    joint_rows.append(
        joint_symmetry_test(
            panel,
            up_col,
            down_col,
            variable_name,
            covariance="date_cluster",
        )
    )

    for lag in [1, 5, 10]:
        joint_rows.append(
            joint_symmetry_test(
                panel,
                up_col,
                down_col,
                variable_name,
                covariance="driscoll_kraay",
                maxlags=lag,
            )
        )

joint_covariance_comparison = pd.DataFrame(joint_rows)

display(
    joint_covariance_comparison[
        [
            "variable",
            "covariance_label",
            "f_stat",
            "p_value",
        ]
    ].round(6)
)

assert joint_covariance_comparison["p_value"].notna().all(), (
    "공동검정 p_value가 아직 계산되지 않았습니다."
)

joint_covariance_comparison.to_csv(
    ROBUST_TABDIR / "robustness_joint_test_covariance.csv",
    index=False,
    encoding="utf-8-sig",
)

print("공동검정 p_value 정상 계산 완료")
```

## D-6. 다중검정 범위 민감도

다음 세 가지를 확인한다.

1. 3년물·곡률 전 산업 공동검정 2개를 함께 보정
2. 곡률 상승 7개·하락 7개를 따로 보정한 결과와 14개 전체 보정 결과 비교
3. 3년물 Wald 7개·곡률 Wald 7개를 따로 보정한 결과와 14개 전체 보정 결과 비교

```python
# 8-1) 공동검정 2개를 같은 검정족으로 보정
joint_multiple_rows = []

for covariance_label, group in joint_covariance_comparison.groupby(
    "covariance_label",
    sort=False,
):
    group = group.copy()
    pvals = group["p_value"].to_numpy()

    group["p_bh_across_2"] = multipletests(
        pvals,
        alpha=BH_ALPHA,
        method="fdr_bh",
    )[1]
    group["p_bonferroni_across_2"] = multipletests(
        pvals,
        alpha=BH_ALPHA,
        method="bonferroni",
    )[1]
    joint_multiple_rows.append(group)

joint_multiple_testing = pd.concat(joint_multiple_rows, ignore_index=True)
joint_multiple_testing.to_csv(
    ROBUST_TABDIR / "robustness_multiple_testing_joint.csv",
    index=False,
    encoding="utf-8-sig",
)

# 8-2) 곡률 개별 계수 14개 보정 범위 비교
curvature_coefficients = coefficient_table(
    panel,
    "d_curv_up",
    "d_curv_down",
    model_name="Curvature_Coef",
)

curvature_coefficients = add_bh_correction(
    curvature_coefficients,
    group_cols=("model", "direction"),
    p_col="p_value",
).rename(
    columns={
        "p_value_bh": "p_bh_split_by_direction",
        "bh_n_tests": "bh_n_split_by_direction",
        "significant_5pct_bh": "sig_split_by_direction",
    }
)

curvature_coefficients["p_bh_all_14"] = multipletests(
    curvature_coefficients["p_value"].to_numpy(),
    alpha=BH_ALPHA,
    method="fdr_bh",
)[1]
curvature_coefficients["sig_bh_all_14"] = (
    curvature_coefficients["p_bh_all_14"] < BH_ALPHA
)

curvature_coefficients.to_csv(
    ROBUST_TABDIR / "robustness_multiple_testing_curvature.csv",
    index=False,
    encoding="utf-8-sig",
)

# 8-3) Y3·곡률 Wald 14개 보정 범위 비교
wald_y3 = industry_wald_table(
    panel,
    "d_y3_up",
    "d_y3_down",
    model_name="Y3_Wald",
)
wald_curvature = industry_wald_table(
    panel,
    "d_curv_up",
    "d_curv_down",
    model_name="Curvature_Wald",
)

wald_family_comparison = pd.concat(
    [wald_y3, wald_curvature],
    ignore_index=True,
).rename(
    columns={
        "p_value_bh": "p_bh_within_model_7",
        "significant_5pct_bh": "sig_within_model_7",
    }
)

wald_family_comparison["p_bh_all_14"] = multipletests(
    wald_family_comparison["p_value"].to_numpy(),
    alpha=BH_ALPHA,
    method="fdr_bh",
)[1]
wald_family_comparison["sig_bh_all_14"] = (
    wald_family_comparison["p_bh_all_14"] < BH_ALPHA
)

wald_family_comparison.to_csv(
    ROBUST_TABDIR / "robustness_multiple_testing_wald_families.csv",
    index=False,
    encoding="utf-8-sig",
)

print("공동검정 2개 보정")
display(
    joint_multiple_testing[
        [
            "variable",
            "covariance_label",
            "p_value",
            "p_bh_across_2",
            "p_bonferroni_across_2",
        ]
    ].round(6)
)

print("곡률 개별 계수: 방향별 7개 vs 전체 14개")
display(
    curvature_coefficients[
        [
            "industry",
            "direction",
            "beta",
            "p_value",
            "p_bh_split_by_direction",
            "p_bh_all_14",
        ]
    ].round(6)
)

print("비대칭 Wald: 모형별 7개 vs 전체 14개")
display(
    wald_family_comparison[
        [
            "model",
            "industry",
            "p_value",
            "p_bh_within_model_7",
            "p_bh_all_14",
        ]
    ].round(6)
)
```

## D-8. 건설업 강건성 종합표

극단값, 기간, 방향 더미 사양의 건설업 결과를 한 표에 모은다.

- `p_value`: 건설업 Wald 원 p값
- `p_value_bh`: 같은 사양에서 7개 산업 Wald 검정에 대한 BH p값
- `robust_bh5`: BH 5% 기준 유지 여부
- `same_coefficient_sign_pattern_as_baseline`: 기존과 같은 계수 부호 패턴인지

```python
master_parts = []

tmp = extreme_construction.copy()
tmp["category"] = "extreme_value"
master_parts.append(
    tmp[
        [
            "category",
            "specification",
            "beta_up",
            "beta_down",
            "f_stat",
            "p_value",
            "p_value_bh",
            "nobs",
        ]
    ]
)

tmp = period_construction.copy()
tmp["category"] = "period"
tmp["specification"] = tmp["model"]
master_parts.append(
    tmp[
        [
            "category",
            "specification",
            "beta_up",
            "beta_down",
            "f_stat",
            "p_value",
            "p_value_bh",
            "nobs",
        ]
    ]
)

tmp = construction_direction_dummy.copy()
tmp["category"] = "direction_intercept"
tmp["specification"] = tmp["model"]
master_parts.append(
    tmp[
        [
            "category",
            "specification",
            "beta_up",
            "beta_down",
            "f_stat",
            "p_value",
            "p_value_bh",
            "nobs",
        ]
    ]
)

construction_master = pd.concat(master_parts, ignore_index=True)
construction_master["robust_bh5"] = construction_master["p_value_bh"] < BH_ALPHA

baseline_construction = extract_industry_row(baseline_y3)
construction_master["same_coefficient_sign_pattern_as_baseline"] = (
    np.sign(construction_master["beta_up"])
    == np.sign(baseline_construction["beta_up"])
) & (
    np.sign(construction_master["beta_down"])
    == np.sign(baseline_construction["beta_down"])
)

construction_master.to_csv(
    ROBUST_TABDIR / "robustness_construction_master.csv",
    index=False,
    encoding="utf-8-sig",
)

display(construction_master.round(6))
```

## E-1. 확장 패널 재확인과 크기 변수 생성

```python
panel = pd.read_csv(config.EXTENDED_PANEL_PATH, encoding="utf-8-sig")
panel["date"] = pd.to_datetime(panel["date"], errors="coerce")

required_panel_cols = [
    "date", "industry", "industry_return", "kospi_return",
    "d_y3", "ns_curvature", "credit_spread",
]
missing_cols = [c for c in required_panel_cols if c not in panel.columns]
if missing_cols:
    raise ValueError("확장 패널에 필요한 열이 없습니다: " + ", ".join(missing_cols))

panel = panel.sort_values(["date", "industry"]).reset_index(drop=True)

assert "credit_spread" in panel.columns, "구버전 패널입니다. data/processed/ 를 갱신하세요."
assert abs(panel["ns_curvature"].mean() + 0.3999) < 0.01, \
    "lambda=0.5 패널이 아닙니다. data/processed/ 를 갱신하세요."

# 방향 변수(STEP7과 동일)와 크기 변수를 함께 만든다.
panel["d_y3_up"] = panel["d_y3"].clip(lower=0)
panel["d_y3_down"] = panel["d_y3"].clip(upper=0)
panel["abs_d_y3"] = panel["d_y3"].abs()

valid = panel["d_y3"].notna()
# 분해 항등식: Δy = Δy_up + Δy_down, |Δy| = Δy_up - Δy_down
assert np.allclose(panel.loc[valid, "d_y3"],
                   panel.loc[valid, "d_y3_up"] + panel.loc[valid, "d_y3_down"])
assert np.allclose(panel.loc[valid, "abs_d_y3"],
                   panel.loc[valid, "d_y3_up"] - panel.loc[valid, "d_y3_down"])
assert (panel["d_y3_up"].dropna() >= 0).all()
assert (panel["d_y3_down"].dropna() <= 0).all()

INDUSTRIES = sorted(panel["industry"].dropna().unique())

print("ns_curvature 평균:", round(panel["ns_curvature"].mean(), 4),
      "| config.NS_LAMBDA =", config.NS_LAMBDA)
print("분석 기간:", panel["date"].min().date(), "~", panel["date"].max().date())
print("산업 수:", len(INDUSTRIES), "| 산업 목록:", INDUSTRIES)
print("변수 생성·검증 완료")
display(panel[["date", "industry", "d_y3", "d_y3_up", "d_y3_down", "abs_d_y3"]].head())
```

## E-2. 재모수화 관계 확립

### 왜 이것부터 확인하나?

절댓값 모형은 STEP7의 상승·하락 모형과 <b>다른 모형이 아니다</b>. 다음 항등식이 성립한다.

$$
\beta_{up}\Delta y^{up} + \beta_{down}\Delta y^{down}
= \beta_{abs}|\Delta y| + \beta_{dir}\Delta y,
\qquad
\beta_{abs} = \frac{\beta_{up}-\beta_{down}}{2},\;
\beta_{dir} = \frac{\beta_{up}+\beta_{down}}{2}
$$

같은 모형을 다르게 적은 것이라 설명력(R²)도 똑같다.

이걸 먼저 확인하는 이유는 두 가지다.

- 이 부록이 더하는 것은 새 자료나 새 모형이 아니라, <b>아직 검정하지 않았던 것</b>과 <b>계수가 실제로 얼마나 큰지</b>라는 점을 분명히 하기 위해서다.
- "크기 계수가 0인가"는 "상승계수와 하락계수가 같은가"와 같은 질문이다. 그래서 STEP7·PART D가 이미 한 검정과 어디까지 겹치는지 미리 짚어둘 수 있다.

```python
def fit_updown(data, hac_lags=HAC_LAGS):
    """R = a + g*kospi + b_up*Δy_up + b_down*Δy_down (HAC). STEP7과 동일 사양."""
    cols = ["industry_return", "kospi_return", "d_y3_up", "d_y3_down"]
    s = data[cols].replace([np.inf, -np.inf], np.nan).dropna()
    X = sm.add_constant(s[["kospi_return", "d_y3_up", "d_y3_down"]], has_constant="add")
    return sm.OLS(s["industry_return"], X).fit(
        cov_type="HAC", cov_kwds={"maxlags": hac_lags})


def fit_absdir(data, hac_lags=HAC_LAGS):
    """R = a + g*kospi + b_abs*|Δy| + b_dir*Δy (HAC). 위 모형의 재모수화."""
    cols = ["industry_return", "kospi_return", "abs_d_y3", "d_y3"]
    s = data[cols].replace([np.inf, -np.inf], np.nan).dropna()
    X = sm.add_constant(s[["kospi_return", "abs_d_y3", "d_y3"]], has_constant="add")
    return sm.OLS(s["industry_return"], X).fit(
        cov_type="HAC", cov_kwds={"maxlags": hac_lags})


rows = []
for ind in INDUSTRIES:
    g = panel[panel["industry"] == ind]
    m1, m2 = fit_updown(g), fit_absdir(g)
    b_abs_implied = (m1.params["d_y3_up"] - m1.params["d_y3_down"]) / 2
    b_dir_implied = (m1.params["d_y3_up"] + m1.params["d_y3_down"]) / 2
    rows.append({
        "industry": ind,
        "beta_up": m1.params["d_y3_up"],
        "beta_down": m1.params["d_y3_down"],
        "beta_abs_implied": b_abs_implied,
        "beta_abs_direct": m2.params["abs_d_y3"],
        "abs_diff": abs(b_abs_implied - m2.params["abs_d_y3"]),
        "beta_dir_implied": b_dir_implied,
        "beta_dir_direct": m2.params["d_y3"],
        "dir_diff": abs(b_dir_implied - m2.params["d_y3"]),
        "r2_updown": m1.rsquared,
        "r2_absdir": m2.rsquared,
        "r2_diff": abs(m1.rsquared - m2.rsquared),
        "nobs": int(m2.nobs),
    })

uncert_reparameterization = pd.DataFrame(rows)

# 항등식이 수치오차 범위에서 성립하는지 확인한다.
assert (uncert_reparameterization["abs_diff"] < 1e-10).all()
assert (uncert_reparameterization["dir_diff"] < 1e-10).all()
assert (uncert_reparameterization["r2_diff"] < 1e-12).all()

uncert_reparameterization.to_csv(
    TABDIR / "uncert_reparameterization.csv", index=False, encoding="utf-8-sig")
print("저장:", config._rel(TABDIR / "uncert_reparameterization.csv"))
display(uncert_reparameterization.round(6))
```

> 이 표는 산업별로 상승·하락 모형에서 유도한 계수와 절댓값 모형에서 직접 추정한 계수를 나란히 놓은 것이다. 두 모형이 같은 모형의 재모수화인지 확인하기 위한 표다.

### 관찰 결과

- 모든 산업에서 `beta_abs_implied`와 `beta_abs_direct`의 차이가 1e-10 미만이다.
- `beta_dir_implied`와 `beta_dir_direct`의 차이도 1e-10 미만이다.
- 두 모형의 R² 차이는 1e-12 미만이다.

## E-3. 산업별 크기·방향 계수와 BH 보정

$$
R_{i,t} = \alpha_i + \gamma_i R_{KOSPI,t}
+ \beta_{abs,i}|\Delta y_{3,t}| + \beta_{dir,i}\Delta y_{3,t} + \epsilon_{i,t}
$$

### 계수를 읽는 법

- 크기 계수가 <b>음수</b>이면, 금리가 어느 쪽으로 가든 많이 움직일수록 그 산업 수익률이 낮아지는 쪽이다.
- 방향 계수는 부호 그대로 읽는다. 양수면 금리가 오를 때 오르는 쪽이다.

### STEP7과 무엇이 다른가

STEP7은 산업별 계수와 원래 p값만 보고했다. 여기서는 7개 산업에 <b>다중검정 보정</b>을 적용한다. 검정을 여러 번 하면 우연히 유의한 게 섞이므로, 크기와 방향을 각각 따로 보정한다.

```python
# add_bh_correction -- nb06(06_yield_curve_exposure.ipynb) cell 42 원문 그대로.
def add_bh_correction(result, group_cols=("model", "factor"),
                      p_col="p_value", alpha=0.05):
    out = result.copy()
    group_cols = list(group_cols)

    out["p_value_bh"] = np.nan
    out["bh_n_tests"] = 0

    # 길이 1 리스트로 그룹화하면 pandas 4에서 키가 스칼라 대신 튜플이 된다. 단일 열이면 스칼라로 넘긴다.
    _by = group_cols[0] if len(group_cols) == 1 else group_cols
    for _, _sub in out.groupby(_by):
        idx = _sub.index
        pvals = out.loc[idx, p_col].to_numpy(dtype=float)
        out.loc[idx, "p_value_bh"] = multipletests(pvals, method="fdr_bh")[1]
        out.loc[idx, "bh_n_tests"] = len(pvals)

    out["bh_family"] = out[group_cols].astype(str).agg(" | ".join, axis=1)
    out["significant_5pct_bh"] = out["p_value_bh"] < alpha

    return out


rows = []
for ind in INDUSTRIES:
    g = panel[panel["industry"] == ind]
    m = fit_absdir(g)
    ci = m.conf_int(alpha=0.05)
    for comp, var in [("size", "abs_d_y3"), ("direction", "d_y3")]:
        # p값은 STEP7·STEP7a와 같은 Wald F 기준을 쓴다. t^2 = F 이므로 같은 검정이며,
        # 참조분포만 달라 t 기준으로 쓰면 소수점 넷째 자리에서 값이 어긋난다.
        rows.append({
            "model": "Y3_SizeDirection",
            "industry": ind,
            "component": comp,
            "beta": m.params[var],
            "se_hac": m.bse[var],
            "t_value": m.tvalues[var],
            "p_value": float(m.f_test(f"{var}=0").pvalue),
            "ci_lower_95": ci.loc[var, 0],
            "ci_upper_95": ci.loc[var, 1],
            "nobs": int(m.nobs),
            "r_squared": m.rsquared,
            "hac_lags": HAC_LAGS,
        })

uncert_size_direction_betas = add_bh_correction(
    pd.DataFrame(rows), group_cols=("model", "component"), alpha=BH_ALPHA)

uncert_size_direction_betas.to_csv(
    TABDIR / "uncert_size_direction_betas.csv", index=False, encoding="utf-8-sig")
print("저장:", config._rel(TABDIR / "uncert_size_direction_betas.csv"))
display(uncert_size_direction_betas.round(6))
```

> 이 표는 산업x성분(14행) 단위의 계수·HAC 표준오차·BH p값을 보여준다. 검정족은 크기 7개·방향 7개로 각각 나눠 보정했다.

> 참고 — STEP7은 크기 계수를 구할 때 방향 항을 빼고 계산했다. 이 부록은 항상 둘 다 넣는다. 방향 항을 빼면 크기 계수가 방향 효과를 조금 흡수하므로, 상승·하락 모형과 정확히 같아지는 것은 둘 다 넣은 쪽이다. 아래 셀에서 이 관계를 확인한다.

```python
# 입력 정합성 확인: STEP7 표(읽기 전용)와 사양이 같은 방향계수가 일치하는지만 본다.
_ref_path = TABDIR / "absolute_value_model.csv"
if _ref_path.exists():
    ref = pd.read_csv(_ref_path, encoding="utf-8-sig")
    new = (uncert_size_direction_betas
           .pivot(index="industry", columns="component", values="beta")
           .reset_index())
    cmp = ref.merge(new, on="industry", how="inner")
    dir_diff = (cmp["beta_dir_both"] - cmp["direction"]).abs().max()
    size_gap = (cmp["beta_abs"] - cmp["size"]).abs().max()
    assert dir_diff < 1e-8, "같은 사양인 방향계수가 일치하지 않습니다."
    print(f"방향계수(사양 동일) 최대 차이: {dir_diff:.2e} -> 일치")
    print(f"크기계수(사양 상이) 최대 차이: {size_gap:.2e} -> 방향항 포함 여부에서 오는 차이")
else:
    print("참조 파일이 없어 확인을 건너뜁니다.")
```

## E-4. 산업 간 크기계수 동일성 검정

$$
H_0:\ \beta_{abs,1} = \beta_{abs,2} = \cdots = \beta_{abs,7}
$$

### STEP7의 검정과 무엇이 다른가

STEP7은 "상승계수가 산업마다 같은가", "하락계수가 산업마다 같은가"를 각각 검정했다.
여기서 묻는 것은 "<b>상승계수와 하락계수의 차이</b>가 산업마다 같은가"다. 앞의 둘 중 어느 것과도 같은 질문이 아니다.

### 표준오차

같은 날 여러 산업이 함께 움직이는데, 산업별로 따로 계산하는 표준오차는 이걸 놓친다.
그래서 산업끼리 비교할 때는 <b>같은 날짜끼리 묶는 표준오차</b>를 쓴다. 프로젝트 표준이다.

```python
# run_industry_difference_test -- nb06 cell 68 원문 그대로
def run_industry_difference_test(data, test_vars, model_name,
                                 ret_col="industry_return",
                                 mkt_col="kospi_return",
                                 industry_col="industry",
                                 date_col="date"):
    """H0: 모든 산업의 계수가 같다. 요인별로 F-검정한다."""
    test_vars = list(test_vars)
    d = data.dropna(subset=[ret_col, mkt_col, industry_col, date_col] + test_vars).copy()

    date_codes = pd.factorize(d[date_col])[0]
    if len(np.unique(date_codes)) < 2:
        raise ValueError("날짜 그룹이 2개 이상이어야 클러스터 SE를 쓸 수 있습니다.")

    dummies = pd.get_dummies(d[industry_col], prefix="ind", dtype=float)
    blocks = [dummies, dummies.mul(d[mkt_col].values, axis=0).add_prefix("mkt_")]

    for v in test_vars:
        blocks.append(dummies.mul(d[v].values, axis=0).add_prefix(f"{v}__"))

    X = pd.concat(blocks, axis=1)
    names = list(X.columns)

    model = sm.OLS(d[ret_col].values, X.values).fit(
        cov_type="cluster", cov_kwds={"groups": date_codes})

    rows = []
    for v in test_vars:
        idx = [i for i, c in enumerate(names) if c.startswith(f"{v}__")]

        # "첫 산업 계수 = 나머지 산업 계수" 제약행렬
        R = np.zeros((len(idx) - 1, len(names)))
        for r, j in enumerate(idx[1:]):
            R[r, idx[0]] = 1.0
            R[r, j] = -1.0

        t = model.f_test(R)

        rows.append({
            "model": model_name,
            "variable": v,
            "f_stat": float(t.fvalue),
            "p_value": float(t.pvalue),
            "df_num": int(len(idx) - 1),
            "n_industries": int(dummies.shape[1]),
            "n_obs": int(model.nobs),
            "n_dates": int(len(np.unique(date_codes))),
            "significant_5pct": bool(t.pvalue < 0.05),
        })

    return pd.DataFrame(rows)


# 크기·방향 두 변수를 한 모형에 함께 넣어 검정한다.
diff_new = run_industry_difference_test(
    panel, ["abs_d_y3", "d_y3"], "Y3_SizeDirection")
# 한 모형에서 두 변수를 검정하므로 STEP7과 같은 방식으로 BH 보정한다
# (07의 industry_difference_tests.csv 도 검정족 2개로 보정했다).
diff_new = add_bh_correction(diff_new, group_cols=("model",), alpha=BH_ALPHA)
diff_new["source"] = "07b (신규)"

# STEP7 기존 결과(읽기 전용)를 참고로 붙인다.
_ref = TABDIR / "industry_difference_tests.csv"
if _ref.exists():
    ref = pd.read_csv(_ref, encoding="utf-8-sig")
    ref = ref[ref["model"] == "Y3_Asymmetry"][
        ["model", "variable", "f_stat", "p_value", "df_num",
         "n_industries", "n_obs", "n_dates", "significant_5pct",
         "p_value_bh", "bh_n_tests", "significant_5pct_bh"]].copy()
    ref["source"] = "07 (기존, 참고)"
    uncert_industry_difference = pd.concat([diff_new, ref], ignore_index=True)
else:
    uncert_industry_difference = diff_new

uncert_industry_difference.to_csv(
    TABDIR / "uncert_industry_difference.csv", index=False, encoding="utf-8-sig")
print("저장:", config._rel(TABDIR / "uncert_industry_difference.csv"))
display(uncert_industry_difference.round(6))
```

> 이 표는 산업 간 계수 동일성 F검정을 보여준다. 위 2행이 이 부록에서 새로 검정한 크기·방향 성분이고, 아래 2행은 STEP7이 이미 보고한 상승·하락 계수 검정을 참고로 붙인 것이다.

### 관찰 결과

- 크기 성분의 산업 간 동일성 검정은 F(6)=2.2129, 원 p=0.0392, BH p=0.0392로 관찰되었다.
- 방향 성분의 산업 간 동일성 검정은 F(6)=4.1372, 원 p=0.0004, BH p=0.0008로 관찰되었다.
- STEP7이 보고한 상승계수 동일성(F=3.5162, p=0.0018)·하락계수 동일성(F=3.1304, p=0.0047)과 비교하면, 크기 성분의 p값이 더 크게 관찰되었다.

## E-5. 조건을 바꿔도 산업 간 차이가 남는가

5단계 결과가 특정 시기나 특정 날짜에만 기대고 있는지 본다.

### PART D와 겹치지 않게 하는 것

PART D는 <b>건설업 하나</b>에 대해 조건별 민감도를 이미 봤다.
그때의 검정("상승계수와 하락계수가 같은가")은 "건설업 크기계수가 0인가"와 같은 검정이라,
여기서 다시 계산하면 PART D와 같은 표가 된다. 그래서 <b>건설업 개별 검정은 싣지 않는다</b>.

여기서 보는 것은 PART D가 보지 않은 <b>산업 간 크기계수 차이</b>가 조건에 따라 어떻게 달라지는지 하나다.

### 조건 목록

PART D가 쓴 목록을 그대로 따른다. 다른 조건에서는 결과가 달라질 수 있다.
건설업 제외 조건은 PART D에 없던 것으로, 관찰된 차이가 특정 산업에서 오는지 보기 위해 넣는다.

```python
def spec_row(data, label, note):
    """조건 하나에 대해 산업 간 크기계수 차이 검정을 돌린다."""
    diff = run_industry_difference_test(data, ["abs_d_y3", "d_y3"], label)
    row_abs = diff[diff["variable"] == "abs_d_y3"].iloc[0]
    row_dir = diff[diff["variable"] == "d_y3"].iloc[0]

    return {
        "specification": label,
        "note": note,
        "n_obs": int(row_abs["n_obs"]),
        "n_dates": int(row_abs["n_dates"]),
        "n_industries": int(row_abs["n_industries"]),
        "size_diff_f": row_abs["f_stat"],
        "size_diff_p": row_abs["p_value"],
        "size_keeps_5pct": bool(row_abs["p_value"] < 0.05),
        "direction_diff_f": row_dir["f_stat"],
        "direction_diff_p": row_dir["p_value"],
        "direction_keeps_5pct": bool(row_dir["p_value"] < 0.05),
    }


base = panel.dropna(subset=["industry_return", "kospi_return", "d_y3"]).copy()
year = base["date"].dt.year

# 극단일 기준은 날짜 단위(중복 제거)로 계산한다.
_daily_abs = base.drop_duplicates("date")["abs_d_y3"]
q99, q995 = _daily_abs.quantile(0.99), _daily_abs.quantile(0.995)

wins = base.copy()
_lo, _hi = wins["d_y3"].quantile([0.01, 0.99])
wins["d_y3"] = wins["d_y3"].clip(_lo, _hi)
wins["abs_d_y3"] = wins["d_y3"].abs()

specs = [
    (base, "전체 2016-2025", "기준 사양"),
    (base[~year.isin([2022, 2023])], "2022-2023 제외", "금리 급변동기 제외"),
    (base[~year.isin([2020, 2021])], "2020-2021 제외", "코로나 구간 제외"),
    (base[year <= 2020], "2016-2020", "전반기"),
    (base[year >= 2021], "2021-2025", "후반기"),
    (base[base["abs_d_y3"] < q99], "절대변화 상위1% 제외", "극단일 제외"),
    (base[base["abs_d_y3"] < q995], "절대변화 상위0.5% 제외", "극단일 제외"),
    (wins, "Winsorize 1%", "상하위 1% 절단"),
    (base[base["industry"] != "건설업"], "건설업 제외", "특정 산업 의존 확인"),
]

uncert_specification_sensitivity = pd.DataFrame(
    [spec_row(d, lab, note) for d, lab, note in specs])

uncert_specification_sensitivity.to_csv(
    TABDIR / "uncert_specification_sensitivity.csv", index=False, encoding="utf-8-sig")
print("저장:", config._rel(TABDIR / "uncert_specification_sensitivity.csv"))
display(uncert_specification_sensitivity.round(6))
```

> 이 표는 조건별로 산업 간 크기계수 차이 검정의 결과를 보여준다. 비교 기준으로 방향계수 차이도 함께 실었다.

### 관찰 결과

- 크기계수 차이가 5% 기준을 넘은 조건: 전체(0.0392), 2022-2023 제외(0.0162), 2020-2021 제외(0.0210), 2021-2025(0.0061), Winsorize 1%(0.0327).
- 넘지 않은 조건: 2016-2020(0.3195), 절대변화 상위 1% 제외(0.0542), 상위 0.5% 제외(0.0744), 건설업 제외(0.4469).
- 방향계수 차이는 2022-2023 제외(0.2539)와 2016-2020(0.3272)에서 5% 기준을 넘지 않았고, 나머지 조건에서는 넘었다.
- 극단일을 뺀 조건에서 두 검정의 결과가 갈렸다. 크기 쪽은 5% 기준을 넘지 않았고(0.0542, 0.0744), 방향 쪽은 넘었다(둘 다 0.0000).
- 2022-2023을 뺀 조건에서도 두 검정의 결과가 갈렸다. 크기 쪽은 0.0162, 방향 쪽은 0.2539로 관찰되었다.

## E-6. 경제적 크기 환산

### 유의성과 크기는 다른 질문이다

계수가 5% 기준을 만족하는지와, 그 계수가 나타내는 변화가 실제로 큰지는 다른 질문이다.
여기서는 두 가지 기준으로 환산한다.

- $|\Delta y_3|$ 가 1 표준편차 움직였을 때의 수익률 변화를, 해당 산업 <b>일별 수익률 표준편차 대비 비율</b>로 표시한다.
- 같은 기준으로 KOSPI 1 표준편차 효과를 함께 계산해 비교 기준으로 둔다.

```python
rows = []
for ind in INDUSTRIES:
    g = panel[panel["industry"] == ind]
    m = fit_absdir(g)
    s = g[["industry_return", "kospi_return", "abs_d_y3", "d_y3"]].dropna()

    ret_sd = s["industry_return"].std()
    abs_sd = s["abs_d_y3"].std()
    kospi_sd = s["kospi_return"].std()

    b_abs = m.params["abs_d_y3"]
    rows.append({
        "industry": ind,
        "beta_abs": b_abs,
        "p_value": m.pvalues["abs_d_y3"],
        "abs_d_y3_sd_pp": abs_sd,
        "abs_d_y3_sd_bp": abs_sd * 100,
        "ret_sd_pct": ret_sd * 100,
        # 1SD 충격 시 수익률 변화(%)
        "effect_1sd_pct": b_abs * abs_sd * 100,
        # 수익률 표준편차 대비 비율(%)
        "ratio_to_ret_sd_pct": abs(b_abs * abs_sd) / ret_sd * 100,
        # 10bp 변동 시 수익률 변화(%)
        "effect_10bp_pct": b_abs * 0.1 * 100,
        # 비교 기준: KOSPI 1SD 효과
        "kospi_effect_1sd_pct": m.params["kospi_return"] * kospi_sd * 100,
        "kospi_ratio_to_ret_sd_pct": abs(m.params["kospi_return"] * kospi_sd) / ret_sd * 100,
        "nobs": int(m.nobs),
    })

uncert_economic_magnitude = pd.DataFrame(rows)
uncert_economic_magnitude.to_csv(
    TABDIR / "uncert_economic_magnitude.csv", index=False, encoding="utf-8-sig")
print("저장:", config._rel(TABDIR / "uncert_economic_magnitude.csv"))
display(uncert_economic_magnitude.round(4))
```

> 이 표는 |Δy3| 1 표준편차 충격의 수익률 효과를 산업별 수익률 표준편차 대비 비율로 환산하고, KOSPI 1 표준편차 효과를 비교 기준으로 함께 보여준다.

### 관찰 결과

- |Δy3| 1 표준편차 충격의 효과는 산업 일별 수익률 표준편차의 0.04%~4.34% 범위로 관찰되었다.
- 같은 기준으로 KOSPI 1 표준편차 효과는 48.8%~85.0% 범위로 관찰되었다.
- 건설업의 10bp 환산 효과는 −0.263%로 관찰되었다.

### PART E 해석

<b>원래 예상은 이랬다.</b> 빚이 많은 기업은 금리가 오르면 이자 부담과 대출 축소를 함께 겪는다. 반대로 금리가 내린다고 그만큼 좋아지지는 않는다. 그래서 오를 때와 내릴 때 반응이 다를 것으로 봤다.

<b>건설업 결과는 이 그림과 다르다.</b> 금리가 오를 때도 내릴 때도 수익률이 낮아지는 쪽으로 나온다. 방향보다 <b>얼마나 움직였는지</b>와 더 이어져 보인다.

<b>이 모습은 실물옵션 논의와 방향이 맞는다</b>(Bernanke 1983, Dixit & Pindyck 1994). 되돌리기 어려운 큰 투자를 앞두면, 앞날이 불확실할 때 결정을 미루는 편이 낫다는 논의다. 분양 일정과 자금 조달이 금리 전망에 묶여 있는 건설업에 들어맞을 수 있다.

<b>다만 이 설명이 맞는지는 지금 자료로 확인할 수 없다.</b> 착공·분양 일정도, 자금 조달 규모도, 산업별 부채 구조도 자료에 없다. 방향이 맞는다는 정도까지만 말할 수 있다.

<b>산업 간 차이는 방향 쪽이 더 뚜렷하다.</b> 크기 쪽 p값(0.0392)이 방향 쪽(0.0004)보다 크다. 크기가 방향보다 산업을 더 잘 가른다고 볼 근거는 없다.

<b>건설업을 빼면 산업 간 차이가 남지 않는다</b>(p=0.4469). "산업마다 다르다"보다 <b>"건설업이 나머지와 다르다"</b>에 가까운 모습이다. STEP7도 크기 반응이 건설업에만 있다고 적었지만 검정은 하지 않았다. 이번 결과는 그 서술과 방향이 같다.

<b>조건을 바꿔도 대체로 남지만, 급변동일과 초반 기간에는 남지 않는다.</b> 금리가 크게 움직인 날을 빼거나 2016-2020년만 보면 5% 기준을 넘지 않는다. 결과가 최근 시기와 급변동일에 상당히 기대고 있을 수 있다.

<b>크기와 방향은 같은 민감도 패턴을 보이지 않는다.</b> 두 검정은 서로 다른 제거 조건에서 서로 다른 강건성을 보였다. 극단일을 뺀 조건에서는 크기 쪽만 5% 기준을 넘지 않았고, 2022-2023을 뺀 조건에서는 방향 쪽만 넘지 않았다. 여기서 더 나아가 두 축이 서로 다른 정보원에서 나온다고까지 말할 근거는 이 표본에서 관찰되지 않는다.

### PART E 한계

- 절댓값 모형은 상승·하락 모형을 다르게 적은 것이다. 새 자료나 새 모형을 더한 게 아니라, 아직 검정하지 않았던 것을 검정하고 계수 크기를 환산했다.
- 여기서 쓴 변동 크기는 <b>그날 실제로 움직인 폭</b>이다. 앞날에 대한 불안을 직접 잰 값이 아니다. 국내에는 그런 지표를 만들 시장 자료가 없다. 그래서 "불확실성에 반응한다"는 표현은 해석에 그친다.
- KOSPI를 걷어내고 남은 <b>같은 날 움직임의 관계</b>다. 금리가 주가를 움직였다는 뜻은 아니다.
- 산업 단위 자료라 기업별 차이는 볼 수 없다. 실물옵션 설명을 검정하려면 기업 부채·투자 자료가 필요하고, 이는 H3에서 다룰 몫이다.
- 금리가 크게 움직인 날을 빼면 5% 기준을 넘지 않는다. 결과가 급변동일에 상당히 기대고 있을 수 있다.
- 2016-2020년만 보면 5% 기준을 넘지 않는다. 전 기간에 공통된 구조로 보기는 어렵다.
- 조건 목록은 PART D가 쓴 것을 그대로 따랐다. 다른 조건에서는 결과가 달라질 수 있다.
- 크기 계수는 방향 항을 넣느냐에 따라 값이 달라진다. STEP7이 적은 값과 이번 값은 서로 다른 식에서 나온 것이므로 같은 숫자로 인용하지 않는다.

### PART E Takeaway

- 절댓값 모형은 STEP7 상승·하락 모형을 다르게 적은 것이다. 이 부록이 더한 것은 아직 검정하지 않았던 것과 계수 크기 환산이다.
- 다중검정 보정 후 크기 계수가 5% 기준을 넘은 산업은 건설업 하나다.
- 산업 간 크기계수 차이는 p=0.0392로, 방향계수 차이(p=0.0004)보다 약하게 나왔다.
- 건설업을 빼면 산업 간 차이가 5% 기준을 넘지 않는다.
- 금리 변동 크기의 영향은 산업 하루 등락폭의 5% 미만이다. 같은 기준으로 KOSPI는 49~85%다.
- 실물옵션 해석은 결과와 방향이 맞는 수준이며, 지금 자료로 이 설명을 검정할 수는 없다.

# 부록 A. 어느 표에 어떤 표준오차를 왜 썼는가

표준오차는 가설(H1/H2)이 아니라 <b>회귀 형태</b>를 기준으로 정한다. 가설 단위로 정하면 같은 형태의 회귀에 서로 다른 SE가 붙어 기준이 흐려지기 때문이다.

- <b>통합회귀</b> — 산업더미 교차항으로 7개 산업이 한 회귀식에 함께 들어간다. 같은 날짜의 산업들이 공통 충격을 받아 잔차가 횡단면으로 상관되므로 <b>날짜 클러스터</b> SE를 쓴다. 시계열 HAC는 이 상관을 보정하지 못한다.

# 부록 B. BH(FDR) 보정을 어떤 단위로 묶었는가

BH는 하나의 검정족(family) 안에서 위양성 비율(FDR)을 통제하는 절차이고, 검정족은 <b>연구질문 단위</b>로 정의한다.

- Baseline(산업 간 차이)·H1(곡선요인)·H2(비대칭)은 서로 다른 연구질문이므로 각각 별도 family로 둔다.
- 전체를 하나로 일괄 보정하면, H1에서 검정을 몇 개 했는지가 Baseline 표의 임계값을 바꾸게 된다. 그러면 Baseline 결과가 왜 이렇게 나왔는지를 Baseline 가설 안에서 설명할 수 없다.
- 다만 이 선택이 결과를 유리하게 만든 것이 아님을 보이기 위해, <b>전체를 일괄 보정했을 때의 q값을 같은 표에 병기</b>한다.

표 읽는 법

- `보정 전 p` — 각 검정의 원 p값.
- `가설 단위 q` — 본문 판정에 쓰는 BH 보정값. 위 family 안에서만 보정한다.
- `전체 일괄 q` — 04·05·06·07의 확증 검정 전체를 하나의 족으로 묶었을 때의 값. 비교용이며 본문 판정에는 쓰지 않는다.
- PART D의 강건성·민감도 검정(`robustness_*`)과 PART E의 확장 검정(`uncert_*`)은 본문 확증 검정과 검정족이 달라 일괄 보정 대상에서 뺀다.

전체 일괄 q는 다른 노트북의 결과표까지 읽어 계산한다. 04·05·06·07을 모두 한 번씩 실행한 뒤에 이 셀을 돌려야 값이 채워진다.

```python
# 부록 B — BH 보정 단위 요약 (family / 검정 수 / 보정 전 p / 가설 단위 q / 전체 일괄 q)

_bh_source_tables = [
    ("up_down_beta_long", up_down_beta_long, "p_value", "p_value_bh", "significant_5pct_bh"),
    ("asymmetry_wald_tests", asymmetry_wald_tests, "p_value", "p_value_bh", "significant_5pct_bh"),
    ("key_factor_asymmetry_tests", key_factor_asymmetry_tests, "p_value", "p_value_bh", "significant_5pct_bh"),
    ("key_factor_asymmetry_wald_tests", key_factor_asymmetry_wald_tests, "p_value", "p_value_bh", "significant_5pct_bh"),
    ("industry_difference_tests", industry_difference_tests, "p_value", "p_value_bh", "significant_5pct_bh"),
]
_bh_id_cols = ["industry", "direction", "variable", "model"]

_bh_rows = []
for _src_name, _src_df, _p_col, _p_bh_col, _sig_bh_col in _bh_source_tables:
    for _, _row in _src_df.iterrows():
        _bh_rows.append({
            "source_table": _src_name,
            "bh_family": _row["bh_family"],
            "bh_n_tests": int(_row["bh_n_tests"]),
            "test_id": " | ".join(f"{c}={_row[c]}" for c in _bh_id_cols if c in _row.index),
            "p_raw": float(_row[_p_col]),
            "q_family": float(_row[_p_bh_col]),
            "significant_family_5pct": bool(_row[_sig_bh_col]),
        })

appendix_bh_long = pd.DataFrame(_bh_rows)
appendix_bh_long["q_global"] = multipletests(
    appendix_bh_long["p_raw"].to_numpy(dtype=float), method="fdr_bh"
)[1]
appendix_bh_long["significant_global_5pct"] = appendix_bh_long["q_global"] < config.BH_ALPHA

def _bh_verdict(fam_sig, glob_sig):
    if fam_sig == glob_sig:
        return "동일"
    return "유의→비유의" if fam_sig else "비유의→유의"

appendix_bh_long["판정 변화"] = [
    _bh_verdict(f, g)
    for f, g in zip(appendix_bh_long["significant_family_5pct"], appendix_bh_long["significant_global_5pct"])
]

appendix_bh_detail = appendix_bh_long[[
    "source_table", "bh_family", "bh_n_tests", "test_id", "p_raw", "q_family", "q_global",
    "significant_family_5pct", "significant_global_5pct", "판정 변화",
]].reset_index(drop=True)

appendix_bh = (
    appendix_bh_long.groupby("bh_family", sort=False)
    .agg(
        검정_수=("bh_n_tests", "first"),
        보정_전_p=("p_raw", lambda s: ", ".join(f"{v:.6f}" for v in s)),
        가설_단위_q=("q_family", lambda s: ", ".join(f"{v:.6f}" for v in s)),
        전체_일괄_q=("q_global", lambda s: ", ".join(f"{v:.6f}" for v in s)),
        가설_단위_유의수=("significant_family_5pct", "sum"),
        전체_일괄_유의수=("significant_global_5pct", "sum"),
        판정_변화수=("판정 변화", lambda s: (s != "동일").sum()),
    )
    .reset_index()
)

appendix_bh_path = TABDIR / "appendix_07_bh_families.csv"
appendix_bh.to_csv(appendix_bh_path, index=False, encoding="utf-8-sig")

display(appendix_bh)
print("저장:", config._rel(appendix_bh_path))

# 검정 단위 상세 — 개별 검정마다 보정 전후 p값을 확인한다.
appendix_bh_detail_path = TABDIR / "appendix_07_bh_tests.csv"
appendix_bh_detail.to_csv(appendix_bh_detail_path, index=False, encoding="utf-8-sig")

_flip = appendix_bh_detail["판정 변화"].value_counts().to_dict()
print("저장:", config._rel(appendix_bh_detail_path))
print("가설 단위 vs 전체 일괄 — 5% 판정 비교:", _flip)

display(appendix_bh_detail.round(6))
```
